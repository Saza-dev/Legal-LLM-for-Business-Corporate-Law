API Keys

```python
load_dotenv()
```

Embedding model from hugging face

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

LLM from GROQ

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="llama3-70b-8192")
```

VectorStore

```python
from langchain_chroma import Chroma
vectorstore = Chroma(persist_directory="chroma_legal_db", embedding_function=embedding_model)
```

Retriver

```python
retriever = vectorstore.as_retriever()
```

Prompt

```python
from langchain_core.prompts import ChatPromptTemplate

system_prompt = (
    "You are an Intelligent chatbot who can draft documents based on the given context. Apply the laws and rules in the context when drafting the documents. If you dont have the knowledge in the paticular area to draft the document say that you dont know. Give the Drafted document as output"
    "\n\n"
    "{context}"
)

prompt = ChatPromptTemplate.from_messages(
    [
        ("system",system_prompt),
        ("human","{input}"),
    ]
)
```

Rag chain

```python
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain

# question answer chain
qa_chain = create_stuff_documents_chain(llm,prompt)

# rag chain
rag_chain = create_retrieval_chain(retriever,qa_chain)
```

```python
response = rag_chain.invoke({"input":"Shareholder Agreement"})
print(response["answer"])
```
