```python
import os
from langchain.document_loaders import PyPDFLoader

pdf_dir = "./legal_docs"
documents = []

for filename in os.listdir(pdf_dir):
    if filename.endswith(".pdf"):
        loader = PyPDFLoader(os.path.join(pdf_dir, filename))
        documents.extend(loader.load())
```

```python
documents
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(chunk_size=1500, chunk_overlap=100)
chunks = splitter.split_documents(documents)
```

```python
len(chunks)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
from langchain_chroma import Chroma
db = Chroma.from_documents(chunks, embedding=embedding_model, persist_directory="chroma_legal_db")
```

```python
db = Chroma(persist_directory="chroma_legal_db", embedding_function=embedding_model)
results = db.similarity_search("laws to start a business")

print(results[3].page_content)
```
